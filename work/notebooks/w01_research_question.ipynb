## 2. The question: decision, action, and cost of a wrong call

### Research question

**Which pages should be reviewed first for a potential content refresh or other content action, based on observable search and engagement signals?**

### Decision

The decision is which pages should receive limited review and optimization attention first.

### Action

A reviewer could inspect the highest-priority pages and decide whether to:

- refresh existing content;
- expand thin content;
- improve CTR-related elements;
- protect a page that is performing well;
- consider consolidation or pruning;
- or monitor the page instead of making a change.

### Cost of a wrong recommendation

A false positive could waste editorial or SEO effort on a page that did not need attention.

A false negative could cause a genuinely useful opportunity to be missed.

Because of these costs, the output should be a **ranked review queue**, not an automatic instruction to change a page.

```python
import pandas as pd
import numpy as np

# Load the starter dataset from the repository
DATA_PATH = "data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(DATA_PATH)

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nFirst 5 rows:")
display(df.head())

# Week 1 discovery: pages with impressions
if "impressions_90d" in df.columns:
    positive_impressions = df[df["impressions_90d"] > 0]

    print("\nRows with impressions_90d > 0:", len(positive_impressions))

    if "search_volume" in df.columns:
        corr = positive_impressions[
            ["search_volume", "impressions_90d"]
        ].corr().iloc[0, 1]

        print(
            "Correlation between search_volume and impressions_90d:",
            round(corr, 3)
        )

# Starter label distribution
if "trend_direction" in df.columns:
    declining_rate = df["trend_direction"].eq("down").mean()

    print(
        "Declining rate:",
        round(declining_rate, 3)
    )

# Useful summary statistics for the chosen lane
summary_cols = [
    col for col in [
        "impressions_90d",
        "sessions_90d",
        "ctr",
        "avg_position",
        "content_age_days",
        "days_since_last_update",
        "word_count"
    ]
    if col in df.columns
]

print("\nSummary statistics:")
display(df[summary_cols].describe().T)
```
